# TP05 — Agentes Inteligentes para Othello

Este notebook demonstra o motor de regras e as estratégias implementadas em Python. Para instalar dependências e executar a interface completa, consulte o [README](README.md). O agente Groq usa uma chamada remota por jogada; a célula correspondente é opcional.


```python
from game import BLACK, WHITE, Othello

partida = Othello()
print("Jogadas legais das pretas:", partida.get_valid_moves(BLACK))

# Executa uma jogada legal, captura peças e atualiza o turno.
jogada = partida.get_valid_moves()[0]
pecas_viradas = partida.make_move(BLACK, *jogada)
print(f"Pretas jogaram {jogada}; peças viradas: {pecas_viradas}")
print("Placar:", partida.get_score())
print("Estado:", {k: v for k, v in partida.get_game_status().items() if k != "history"})
```

## Agente heurístico local

A estratégia avalia posições, cantos, material e mobilidade. Ela não usa a rede e pode ser usada para partidas e experimentos reproduzíveis.


```python
from agents import HeuristicAgent

partida_local = Othello()
decisao = HeuristicAgent().play_turn(partida_local)
print(f"Jogada: ({decisao.row}, {decisao.col}) — {decisao.reason}")
print("Placar após a jogada:", partida_local.get_score())
```

## Agente Groq com LangGraph (opcional)

Configure `GROQ_API_KEY` no ambiente ou informe a chave quando solicitado. Esta célula faz uma chamada à API gratuita do Groq sujeita às cotas e aos limites da conta. A chave não deve ser salva no notebook. O grafo observa o jogo pelas tools, pede uma decisão estruturada, valida a jogada e só então executa a ação.


```python
import getpass
import os
from agents import GroqAgent

api_key = os.getenv("GROQ_API_KEY") or getpass.getpass("Chave Groq (não será exibida): ")
partida_groq = Othello()
decisao = GroqAgent(api_key=api_key).play_turn(partida_groq)
print(f"Groq jogou ({decisao.row}, {decisao.col}): {decisao.reason}")
print("Placar após a jogada:", partida_groq.get_score())
```
